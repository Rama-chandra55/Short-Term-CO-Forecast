# Exploratory Data Analysis (EDA)
This notebook explores the UCI Air Quality dataset, handles missing values, and visualizes the distributions.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sys
sys.path.append('../')
from src.preprocess import download_and_extract_data, load_and_clean_data

# Load data
csv_path = download_and_extract_data('../data')
df = load_and_clean_data(csv_path)
df.head()
```

```python
# Check for missing values after cleaning
print(df.isnull().sum())
```

```python
# Time series plot of CO(GT) which is a proxy for AQI
plt.figure(figsize=(15, 5))
plt.plot(df.index, df['CO(GT)'], label='CO(GT)', color='blue', alpha=0.7)
plt.title('Time Series of Carbon Monoxide Concentration (CO(GT))')
plt.xlabel('Date')
plt.ylabel('CO(GT)')
plt.legend()
plt.show()
```

```python
# Correlation Heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Feature Correlation Heatmap')
plt.show()
```
